### **Task 1, Question 2: Evaluation Criteria - Mean Absolute Percentage Error (MAPE)**

This competition uses **Mean Absolute Percentage Error (MAPE)** to measure how close the predicted car prices are to the actual prices in the test dataset.

$$
MAPE = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100
$$

Where:
- $y_i$ = the actual car price
- $\hat{y_i}$ = the predicted car price
- $n$ = the number of records in the test dataset

MAPE shows the average percentage difference between the predicted and actual prices.
For example, a MAPE of 10% means the model’s predictions are off by about 10% on average.

**Why MAPE is useful**
- It gives an easy to understand percentage value, so even non-technical users can see how accurate the model is.
- It treats overestimation and underestimation equally where both would count as errors
- Meaning it works well when prices vary widely (some cheap cars, some expensive ones).

**Limitation of MAPE**
- It any actual prices are very close to zero, MAPE could become unstable since it divides by the actaul value.
- It may **over-penalise errors** on cheaper cars, since small dollar differences can become large percentages.

```python
## 2.2) Create new features from existing features which contain multiple items of information

import re
import numpy as np

# ---------- Helpers ----------

def _to_float(x):
    try:
        if pd.isna(x): return np.nan
        return float(str(x).strip())
    except Exception:
        return np.nan

def parse_power(s):
    """
    Parse 'power' like "236 hp @ 5,200 RPM" -> power_hp, power_rpm
    """
    if pd.isna(s):
        return pd.Series([np.nan, np.nan], index=["power_hp", "power_rpm"])
    t = str(s).lower().replace(",", "")
    m_hp  = re.search(r"([-+]?\d*\.?\d+)\s*hp", t)
    m_rpm = re.search(r"@\s*([-+]?\d*\.?\d+)\s*rpm", t)
    hp  = _to_float(m_hp.group(1))  if m_hp else np.nan
    rpm = _to_float(m_rpm.group(1)) if m_rpm else np.nan
    return pd.Series([hp, rpm], index=["power_hp", "power_rpm"])

def parse_torque(s):
    """
    Parse 'torque' like "266 lb-ft @ 4,000 RPM" or "350 Nm @ 1500 rpm"
    -> torque_lbft (Nm converted to lb-ft), torque_rpm
    """
    if pd.isna(s):
        return pd.Series([np.nan, np.nan], index=["torque_lbft", "torque_rpm"])
    t = str(s).lower().replace(",", "")
    m_tq  = re.search(r"([-+]?\d*\.?\d+)\s*(lb[-\s]?ft|nm)", t)
    m_rpm = re.search(r"@\s*([-+]?\d*\.?\d+)\s*rpm", t)

    tq = np.nan
    if m_tq:
        val = _to_float(m_tq.group(1))
        unit = m_tq.group(2)
        if "nm" in unit:
            val = val * 0.737562  # Nm -> lb-ft
        tq = val

    rpm = _to_float(m_rpm.group(1)) if m_rpm else np.nan
    return pd.Series([tq, rpm], index=["torque_lbft", "torque_rpm"])

def parse_engine_type(s):
    """
    Extract number of cylinders and layout (Inline / V-type / Horizontally (Boxer)/ W-engine ) from engine_type.
    Examples: "I4 Turbo" -> (4, Inline), "V6" -> (6, V-type), "H4" -> (4, Horizontally (Boxer))
    """
    if pd.isna(s):
        return pd.Series([np.nan, np.nan], index=["engine_cylinders", "engine_layout"])
    text = str(s).upper().strip()

    cyl_map = {"I2":2, "I3":3, "I4":4, "I5":5, "I6":6, "V6":6, "V8":8, "V10":10, "V12":12, "H4":4, "H6":6, "W12":12}
    cylinders = np.nan
    for k, v in cyl_map.items():
        if k in text:
            cylinders = v
            break

    if text.startswith("V"):
        layout = "V-type"
    elif text.startswith("I"):
        layout = "Inline"
    elif text.startswith("H"):
        layout = "Boxer"
    elif text.startswith("W"):
        layout = "W-type"
    else:
        layout = "Other"

    return pd.Series([cylinders, layout], index=["engine_cylinders", "engine_layout"])

def extract_transmission_gears(s):
    """
    Extract numeric gear count from transmission_display.
    - 'Continuously Variable Transmission' or 'CVT' -> 0 (no fixed steps)
    - '8-Speed Automatic' -> 8; '6 Speed Manual' -> 6
    - otherwise NaN
    """
    if pd.isna(s):
        return np.nan
    text = str(s).upper().strip()

    if "CVT" in text or "CONTINUOUSLY VARIABLE" in text:
        return 0

    m = re.search(r"(\d+)\s*[- ]*SPEED", text)
    return int(m.group(1)) if m else np.nan


# ---------- One function to apply all parsers to a DF ----------

def apply_parsing(df: pd.DataFrame) -> pd.DataFrame:
    # Power & torque
    if "power" in df.columns:
        df[["power_hp", "power_rpm"]] = df["power"].apply(parse_power)
    if "torque" in df.columns:
        df[["torque_lbft", "torque_rpm"]] = df["torque"].apply(parse_torque)

    # Engine cylinders & layout
    if "engine_type" in df.columns:
        df[["engine_cylinders", "engine_layout"]] = df["engine_type"].apply(parse_engine_type)

    # Transmission gears (CVT → 0)
    if "transmission_display" in df.columns:
        df["trans_gears"] = df["transmission_display"].apply(extract_transmission_gears)

    return df

# ---------- Run on both datasets ----------

train_clean = apply_parsing(train_clean)
test_clean  = apply_parsing(test_clean)

# Quick preview
cols_to_show = [
    c for c in [
        "power","power_hp","power_rpm",
        "torque","torque_lbft","torque_rpm",
        "engine_type","engine_cylinders","engine_layout",
        "transmission_display","trans_gears"
    ] if c in train_clean.columns
]
train_clean[cols_to_show].head()
```

```python
# ---------- Helper functions ----------
def mode_safe(s: pd.Series):
    m = s.mode(dropna=True)
    return m.iloc[0] if len(m) else np.nan

def impute_numeric_with_hierarchy(train, test, col, hier_groups):
    """Fill numeric NaNs using train medians by (make, model, year) → ... → global."""
    if col not in train.columns:
        return train, test
    gmed = train[col].median()

    # Train fill
    tr = train[col].copy()
    for groups in hier_groups:
        groups = [g for g in groups if g in train.columns]
        if not groups:
            continue
        med = train.groupby(groups, dropna=False)[col].transform("median")
        mask = tr.isna() & med.notna()
        tr.loc[mask] = med.loc[mask]
    tr = tr.fillna(gmed)

    # Test fill using medians learned from train
    te = test[col].copy()
    for groups in hier_groups:
        groups = [g for g in groups if g in train.columns]
        if not groups:
            continue
        med_lookup = train.groupby(groups, dropna=False)[col].median()
        if len(groups) == 1:
            mapped = test[groups[0]].map(med_lookup)
        else:
            key = pd.MultiIndex.from_frame(test[groups])
            mapped = pd.Series(med_lookup.reindex(key).values, index=test.index)
        mask = te.isna() & mapped.notna()
        te.loc[mask] = mapped.loc[mask]
    te = te.fillna(gmed)

    train[col] = tr
    test[col]  = te
    return train, test

def impute_categorical_from_train_mode(train, test, col, fill_value="Unknown"):
    """Fill categorical NaNs using mode from training set."""
    if col not in train.columns:
        return train, test
    train_mode = mode_safe(train[col])
    if pd.isna(train_mode):
        train_mode = fill_value
    train[col] = train[col].fillna(train_mode)
    test[col]  = test[col].fillna(train_mode)
    return train, test

train_filled = train_clean.copy()
test_filled = test_clean.copy()

# ---------- Imputation setup ----------
hier = [
    ("make_name","model_name","year"),
    ("make_name","model_name"),
    ("make_name",),
    (),  # global
]

# --- STEP 1: Impute categorical source columns ---
cat_sources = [
    "engine_layout","wheel_system",
    "fuel_type","interior_color","exterior_color",
    "transmission","body_type","city","franchise_dealer"
]
for col in cat_sources:
    if col in train_filled.columns:
        train_filled, test_filled = impute_categorical_from_train_mode(train_filled, test_filled, col)

# --- STEP 2: Impute true numeric source columns ---
numeric_sources = [
    "horsepower","engine_displacement","torque_lbft","torque_rpm","power_hp", "power_rpm", "mileage",
    "seller_rating","savings_amount","latitude","longitude","engine_cylinders",
    "city_fuel_economy","highway_fuel_economy", "back_legroom", "front_legroom",
    "width","height","wheelbase","maximum_seating"
]
for col in numeric_sources:
    if col in train_filled.columns:
        train_filled, test_filled = impute_numeric_with_hierarchy(train_filled, test_filled, col, hier)

# --- STEP 3: Domain rule — ensure CVT coded as 0 gears and electric (if there is) doesn't has engine ---
for df in (train_filled, test_filled):
    if "trans_gears" in df.columns:
        is_cvt = (
            df["transmission"].astype(str)
              .str.contains("CVT|Continuously Variable", case=False, na=False)
        )
        df.loc[is_cvt, "trans_gears"] = 0
    if "fuel_type" in df.columns:
            is_electric = df["fuel_type"].astype(str).str.contains("Electric", case=False, na=False)
            if "engine_displacement" in df.columns:
                df.loc[is_electric, "engine_displacement"] = 0
            if "engine_cylinders" in df.columns:
                df.loc[is_electric, "engine_cylinders"] = 0
            if "engine_layout" in df.columns:
                df.loc[is_electric, "engine_layout"] = "None"

# --- Impute missing trans_gears using median by transmission type (A/M) ---
for df in (train_filled, test_filled):
    if all(col in df.columns for col in ["trans_gears", "transmission"]):

        # 1. Compute median gear count by transmission type (A/M) from TRAIN data
        median_gears_by_type = (
            train_filled.groupby("transmission")["trans_gears"].median().to_dict()
        )

        # 2. Identify missing trans_gears for cars labeled as 'A' or 'M' in transmission
        mask_auto = (df["transmission"]=='A')& (df["trans_gears"].isna())
        mask_man  = (df["transmission"]=='M')& (df["trans_gears"].isna())

        # 3. Impute by median gears based on transmission type (A = Automatic, M = Manual)
        if "A" in median_gears_by_type:
            df.loc[mask_auto, "trans_gears"] = median_gears_by_type["A"]
        if "M" in median_gears_by_type:
            df.loc[mask_man, "trans_gears"] = median_gears_by_type["M"]

        # 4. Handle CVT and electric car explicitly (no fixed gears)
        df.loc[df["transmission"].str.contains("CVT|Continuously Variable", case=False, na=False), "trans_gears"] = 0
        df.loc[df["fuel_type"].str.contains("electric", case=False, na=False), "trans_gears"] = 0

        # 5. Global fallback — fill any remaining NaN with overall median from train
        df["trans_gears"] = df["trans_gears"].fillna(train_filled["trans_gears"].median())

# ---------- Verify ----------
print("Remaining missing (TRAIN):", int(train_filled.isna().sum().sum()))
print("Remaining missing (TEST): ", int(test_filled.isna().sum().sum()))
```

Missing values in both datasets were imputed by data type, with all statisics **solely from training dataset** to prevent data leakage.

- **<u>Categorical variables</u>** were filled with the **mode** from the training set, ensuring that imputed values reflected the most common category observed in real data. Using only training statistics prevents bias transfer to the test set.
    - Hierarchical mode imputation was avoided because many categorical groups (e.g., specific model-years) may have too few samples to produce reliable modes. A single global mode ensures stable, consistent imputation across both datasets and avoids unseen labels during encoding.
- **<u>Numeric variables</u>** were imputed using **hierarchical group medians**, this method fills NaNs using the most specific statistic available, trying (make, model, year), then falling back to (make, model), (make), and finally the global median. This preserves model-specific characteristics and is robust to outliers.
    - For `trans_gears`, the column combined this approach with domain knowledge. CVT transmissions were explicitly set to 0 gears, while other missing values were imputed using the median for their specific transmission type (A/M), based on training data.
    - To future-proof the pipeline, cases where fuel_type indicated Electric automatically set related engine attributes (`engine_displacement`, `engine_cylinders`, `engine_layout`, `trans_gears`) to zero or “None.”

After verification, no missing values remained. The resulting datasets (train_filled and test_filled) are complete and ready for encoding and scaling in later preprocessing stages.

```python
TARGET = 'price'

# Identify feature roles
def is_binary_series(s):
    vals = pd.unique(s.dropna())
    return set(vals).issubset({0,1})

all_cols = [c for c in train_encoded.columns if c != TARGET]
num_cols = train_encoded[all_cols].select_dtypes(include=[np.number]).columns.tolist()

# Binary / one-hot (keep unscaled)
binary_like = [c for c in num_cols if is_binary_series(train_encoded[c])]

# Continuous candidates (numeric minus binary-like; exclude target)
continuous = list((pd.Index(num_cols).difference(binary_like)).difference([TARGET]))
```

```python
# Define columns identified as having high skew or extreme values
cap_cols = [
    'daysonmarket','mileage','savings_amount',
    'engine_displacement','horsepower','power_hp',
    'maximum_seating','torque_lbft','power_rpm','back_legroom'
]

# Filter for columns that actually exist in the dataframes
present = [c for c in cap_cols if c in train_encoded.columns]

# Define a function for IQR-based capping
def iqr_cap(series, factor=1.5):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - factor * iqr, q3 + factor * iqr
    return series.clip(lower, upper), (lower, upper)


caps = {} # Dictionary to store calculated IQR caps for each column

# Create copies of the dataframes
train_capped = train_encoded.copy()
test_capped = test_encoded.copy()

for c in present:
    s = train_capped[c].dropna()
    if s.empty:
        continue
    train_capped[c], (lo, hi) = iqr_cap(s)
    caps[c] = (lo, hi)

# Apply same caps to TEST
for c, (lo, hi) in caps.items():
    if c in test_capped.columns:
        test_capped[c] = test_capped[c].clip(lo, hi)

# Median imputation to the outliers (robust to outliers)
for df in [train_capped, test_capped]:
    for c in present:
        if c in df.columns:
            median_val = train_capped[c].median()  # fit on train
            df[c] = df[c].fillna(median_val)
```

Before model training, several refinement steps were performed to ensure data quality and to prepare the dataset for both tree-based and linear algorithms.

    - `vin` (a unique vehicle ID)

    - The same caps were applied to the test set for consistency, followed by median imputation to handle any remaining missing values.
    - This approach preserves the overall data distribution while minimizing distortion from extreme observations.

| ------------ | -------------------------------- | -------------------------- |

| **Linear models** (`train_linear`, `test_linear`) | Models such as Linear Regression, Ridge| Continuous variables were standardized (mean = 0, std = 1) using parameters learned from the training data. <br> Binary and one-hot features were left unscaled. |

5) Feature Scaling (Standardization for Linear Models)
- Continuous variables were standardized using StandardScaler (z-score) to align feature scales.

6) Dimentionality reduction with PCA

- PCA was fit on the training set and applied to validation/test sets using the same transformation to reduce multicollinearity while preserving most information.
- The 90% variance threshold was chosen to keep most of the meaningful information while reducing noise and redundant patterns among correlated features. This balance helps simplify the dataset without losing important signals, improving model efficiency and reducing the risk of overfitting caused by multicollinearity.

### **Task 3, Question 2**

For this competition, we selected three regression models — **Ridge Regression, Random Forest Regressor, and K-Nearest Neighbors (KNN) Regressor** — to predict used car prices. Model performance is evaluated using Mean Absolute Percentage Error (MAPE), which measures prediction accuracy in percentage terms.

- **Ridge Regression (L2 Regularised Linear Model)** -  a linear model that adds an L2 penalty to reduce large coefficients and control overfitting. It performs well when features are correlated, such as horsepower, torque, and engine size, which often overlap in information. Ridge provides a strong and interpretable baseline model for understanding how each feature influences car price.
- **K-Nearest Neighbors (KNN) Regressor** - a distance-based model that predicts a car’s price based on the average of its k most similar cars. It effectively captures local patterns — for example, cars with similar brand, year, and mileage tend to have similar prices. KNN does not assume any data distribution, but it requires scaled features for accurate distance measurement. The best number of neighbors (k) and distance metric are selected through cross-validation.
- **Random Forest Regressor** - an ensemble of decision trees that combines multiple models to improve prediction accuracy. It captures non-linear relationships and interactions between variables automatically. The model is robust to outliers and identifies which features, such as mileage, brand, or year, have the greatest effect on price. Its averaging process also reduces the risk of overfitting.

However, the **Gradient Boosting Regressor (GBR)** was later implemented as an improvement over the Random Forest model, using sequential tree building to correct previous errors and achieve stronger predictive performance.

#### Suggested Improvement Strategy

From task3 Q.4, the Random Forest model showed stable results (CV MAPE 13.00% vs. Kaggle 9.665%), but the feature importance and error plots indicated that adding more trees or increasing depth no longer improved performance. This suggested that the model had reached its limit in learning patterns. We believed this happened because Random Forests average many independent trees but don’t learn from previous errors, which can limit their ability to reduce bias further.

**Suggestion**: To address this, we replaced the Random Forest with a Gradient Boosting Regressor (GBR) which is an ensemble that builds trees sequentially, where each new tree corrects the errors of the previous ensemble.

**Reasoning**: Unlike Random Forests, Gradient Boosting focuses on sequentially reducing bias and capturing deeper feature interactions. This makes it especially effective for structured datasets like car listings, where relationships are complex but follow clear business logic.

The model’s feature importance results confirmed that it generalised well to real-world car pricing patterns:

- Performance features (e.g., torque, power, rpm) had the strongest positive influence — higher-performance cars command higher prices.
- Age and usage indicators (year, mileage) showed negative relationships — newer, low-mileage vehicles are valued more.
- Physical size and comfort attributes (wheelbase, width, seating) contributed moderately — larger vehicles often justify higher prices.
- Listing and dealer factors (days on market, seller rating) had smaller effects but still reflected buyer perception and trust.

For this evidence, the Gradient Boosting model successfully learned the economic logic of the used-car market, supporting our hypothesis in part EDA earilier that vehicle capability and condition are the main drivers of price, while seller characteristics play a secondary role.
